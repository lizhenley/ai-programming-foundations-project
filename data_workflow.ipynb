## 1. Data Ingestion

This section loads the red and white wine quality datasets from the project repository and combines them into a single DataFrame. A wine type column is added to distinguish between the two datasets.

```python
# Load datasets directly from GitHub
base_url = "https://raw.githubusercontent.com/lizhenley/ai-programming-foundations-project/development/"

red_wine = pd.read_csv(base_url + "winequality-red.csv", sep=";")
white_wine = pd.read_csv(base_url + "winequality-white.csv", sep=";")

# Identify wine types
red_wine["wine_type"] = "Red"
white_wine["wine_type"] = "White"

# Combine datasets
df = pd.concat([red_wine, white_wine], ignore_index=True)

# Display first five rows
df.head()
```

```python
print("Red wine observations:", len(red_wine))
print("White wine observations:", len(white_wine))
print("Combined dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```
